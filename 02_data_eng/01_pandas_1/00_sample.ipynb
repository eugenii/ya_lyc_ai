### Задачи для практики из урока

- 10 самых старых фильмов.
- Сколько фильмов вышло в 1994 году.
- Фильмы Франции с рейтингом выше 8.0.
- Фильмы с названием длиннее 15 символов.
- Сколько фильмов снято в каждой стране (value_counts).
- Топ-5 режиссёров по числу фильмов в рейтинге.
  
И ещё немного практики:

- Три года, в которые вышло больше всего фильмов.
- Фильмы, в описании которых встречается слово «любовь».
- Средний рейтинг фильмов до 1980 года и после 2000 года (результат — обе величины — сохранить и вывести с точностью до двух знаков запятой для удобства).

```python
import pandas as pd
```

```python
df = pd.read_csv('../kinopoisk-top250.csv')
display(df.describe())  
display(df.info())
```

Десять самых старых фильмов


```python
ten_eldest = df.sort_values('year')[['movie', 'year']][:10]
display(ten_eldest)
```

Сколько фильмов вышло в 1994 году.

```python
films_in_1994 = df[df['year'] == 1994]
cnt = films_in_1994.shape[0]
display(cnt)
```

Фильмы Франции с рейтингом выше 8.0.

```python
france_good_films = df[(df['country'] == 'Франция') & (df['rating'] > 8.0)]
display(france_good_films)
```

Фильмы с названием длиннее 15 символов.

```python
long_name_movies = df[df['movie'].str.len() > 15]
display(long_name_movies['movie'])
```

Сколько фильмов снято в каждой стране (value_counts)

```python
grouped_by_country = df.groupby('country')['movie'].count()
display(grouped_by_country)
```

Топ-5 режиссёров по числу фильмов в рейтинге

```python
# display(df.head(2))
directors = df.groupby('director').count()
display(directors.sort_values(by='movie', ascending=False)['movie'])
```

И ещё немного практики:

- Три года, в которые вышло больше всего фильмов.
- Фильмы, в описании которых встречается слово «любовь».
- Средний рейтинг фильмов до 1980 года и после 2000 года (результат — обе величины — сохранить и вывести с точностью до двух знаков запятой для удобства).

```python
max_movies_years = df.groupby('year')['movie'].count()    # .idxmax()
display(max_movies_years.sort_values(ascending=False).head(5))
```

```python
movies_about_love = df[df['overview'].str.contains('любовь')]
display(movies_about_love['movie'])
```

```python
avg_rating_before_1980 = df[df['year'] < 1980]['rating_ball'].mean()
avg_rating_after_2000 = df[df['year'] > 2000]['rating_ball'].mean()
display(round(avg_rating_before_1980, 2), round(avg_rating_after_2000, 2))
```
