- 10 самых старых фильмов.

- Фильмы с названием длиннее 15 символов.

- Топ-5 режиссёров по числу фильмов в рейтинге.

И ещё немного практики:

- Три года, в которые вышло больше всего фильмов.
- Фильмы, в описании которых встречается слово «любовь».
- Средний рейтинг фильмов до 1980 года и после 2000 года (результат — обе величины — сохранить и вывести с точностью до двух знаков запятой для удобства).

```python
import pandas as pd
```

```python
df = pd.read_csv('../kinopoisk-top250.csv')
display(df.describe())  
display(df.info())
```

Десять самых старых фильмов


```python
ten_eldest = df.sort_values('year')[['movie', 'year']][:10]
display(ten_eldest)
```

Сколько фильмов вышло в 1994 году.

```python
films_in_1994 = df[df['year'] == 1994]
cnt = films_in_1994.shape[0]
display(cnt)
```

Фильмы Франции с рейтингом выше 8.0.

```python
france_good_films = df[(df['country'] == 'Франция') & (df['rating'] > 8.0)]
display(france_good_films)
```
