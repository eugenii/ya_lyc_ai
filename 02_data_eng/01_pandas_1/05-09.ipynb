### 8 Рейтинг выше 8.77

Найдите фильмы, которые получили рейтинг выше, чем 8.77. Просуммируйте года производства этих фильмов, чтобы получить ответ.

Например, если у вас 2 фильма – 1+1 (2011) и Побег из Шоушенка (1994) ответом на эту задачу будет 2011 + 1994 = 4005

Формат ввода
Одно число – количество фильмов, вышедшее в 2000-м году.

Формат вывода
Формат вывода – единственное целое число – сумма значений годов, в которые высокорейтинговые фильмы были выпущены в свет.

```python
# Рейтинг выше 8.77

df_top = df[df['rating_ball'] > 8.77]
df_top.head(2)
df_top['year'].count()
display(df_top['year'], df_top['year'].sum())
```

### 9 На рубеже веков
Определите, сколько фильмов вышло в 2000-м году.

Формат ввода
Одно число – количество фильмов, вышедшее в 2000-м году.

Формат вывода
Формат вывода – единственное float число – средний рейтинг среди всех фильмов, округленный до 3 знаков после запятой. Например: 5.678

```python
# количество фильмов в 2000 году

amount_of_films = len(df[df['year'] == 2000])
print("Year 2000 -  amount of films:{}".format(amount_of_films))
avg_raiting = df[df['year'] == 2000]['rating_ball'].mean()
print("Average raitinig of this films: {:.3f}".format(avg_raiting))
```
