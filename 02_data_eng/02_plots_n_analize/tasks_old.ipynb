### № 3
Найти режиссёра, который срежиссировал больше всего фильмов в подборке.
Посчитать средний рейтинг (rating_ball) фильмов этого режиссёра.
Если несколько режиссёров имеют одинаковое максимальное количество фильмов, выберите того, чьё имя минимально в лексикографическом порядке строк Python (то есть min() по строкам).

Формат вывода
Одно вещественное число (float) — средний рейтинг фильмов выбранного режиссёра, округлённый до 3 знаков после запятой:

```python
most_frequent_director = data['director'].value_counts().idxmax()
display(most_frequent_director)
avg_rating = data[data['director'] == most_frequent_director]['rating_ball'].mean()
print(round(avg_rating, 3))
```

### № 4 

Необходимо определить, какой актёр встречается в фильмах подборки чаще всего.

Формат вывода
Имя актёра. Оно может состоять из одного или нескольких слов, разделенных пробелом. Регистр и имя – в точности как в датасете, например Леонардо ДиКаприо

```python
import pandas as pd

PATH = "../kinopoisk-top250.csv"
df = pd.read_csv(PATH)

vc = df['director'].str.strip().value_counts()

# явная сортировка: сначала count (desc), затем director (asc)
top5 = (
    vc.rename_axis('director')
      .reset_index(name='count')
      .sort_values(['count', 'director'], ascending=[False, True])
      .head(5)
)


surnames = top5['director'].str.split().str[-1]
print(",".join(surnames))
```

```python
# Разделяем имена сценаристов и создаем новый DataFrame
screenwriters_split = data['screenwriter'].str.split('; ').explode()

# Подсчитываем количество фильмов для каждого сценариста
screenwriter_counts = screenwriters_split.value_counts()

display(screenwriters_split.head(6))
display(screenwriter_counts.head(6))
```

```python
# Шаг 2: Выберем топ-5 сценаристов
# Теперь нам нужно выбрать топ-5 сценаристов по количеству фильмов. 
# Если несколько сценаристов имеют одинаковое количество фильмов, 
# они должны быть упорядочены в лексикографическом порядке.
# Сортируем сначала по количеству фильмов (по убыванию), 
# затем по имени (в алфавитном порядке)
top_screenwriters = screenwriter_counts.sort_values(ascending=False).head(5)
display(top_screenwriters)
```

```python
# Шаг 3: Для каждого сценариста найдем самый популярный фильм
# Теперь для каждого из топ-5 сценаристов нужно найти фильм с лучшим местом в рейтинге (rating). 
# Это означает, что мы ищем минимальное значение rating среди фильмов,
# связанных с этим сценаристом.
# Сначала создадим новый столбец, который будет содержать список
# сценаристов для каждого фильма:
# Создаем новый столбец с разбитыми именами сценаристов

data['screenwriters_list'] = data['screenwriter'].str.split('; ')
display(data['screenwriters_list'].head(7))
display(data.info())
```

```python
# Теперь для каждого сценариста найдем фильм с минимальным значением rating:
# Список для хранения результатов
most_popular_movies = []

for screenwriter in top_screenwriters.index:
    # Фильтруем фильмы, где этот сценарист указан
    movies_by_screenwriter = data[
        data['screenwriters_list'].apply(lambda x: screenwriter in x)
    ]
    
    # Находим фильм с минимальным значением rating
    most_popular_movie = movies_by_screenwriter.loc[
        movies_by_screenwriter['rating'].idxmin()
    ]
    
    # Добавляем результат в список
    most_popular_movies.append(most_popular_movie)

# Создаем DataFrame из результатов
most_popular_movies_df = pd.DataFrame(most_popular_movies)
display(most_popular_movies_df[['screenwriter', 'movie', 'rating', 'rating_ball']])
```

```python
# Шаг 4: Посчитаем сумму рейтингов rating_ball
# Теперь, когда у нас есть пять самых популярных фильмов, мы можем посчитать сумму их рейтингов
# (rating_ball):
# Считаем сумму рейтингов
total_rating_ball = most_popular_movies_df['rating_ball'].sum()
print(f"Сумма рейтингов (rating_ball) для пяти фильмов: {total_rating_ball}")
```
