### 02 Не самые плодотворные страны

```python
# Импорт библиотек и данных
import pandas as pd

df = pd.read_csv('../kinopoisk-top250.csv')
```

```python
# Обзор данных
df.info()
```

```python
df.head(2)
```

```python
# # Группируем по странам и подсчитываем количество их фильмов

countries_ds = df.groupby('country')['movie'].count()

# Сортируем по убыванию (хотя и не нужно)
countries_ds = countries_ds.sort_values(ascending=False)
print(countries_ds)

# Фильтруем по количеству фильмов и находим сумму
result = countries_ds[countries_ds < 15].sum()
print(f"\nСуммарное количество фильмов в не самых плодотворных странах: {result}")
```
