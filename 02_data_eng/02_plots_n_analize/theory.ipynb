```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../kinopoisk-top250.csv")
df.info()
```

```python
grouped = df.groupby(['year']).agg({
    'rating': ['count', 'mean'],     # количество и средний rating
    'rating_ball': 'mean'            # средний rating_ball
}).round(2)

print(grouped)
```

Гистограмма

```python
years = df['year']
years.head()

plt.hist(years, bins=15)

plt.show()
```

Столбчатая диаграмма

```python
languages = ["Python", "C++", "Java", "JavaScript"]
students_count = [15, 4, 3, 6]
plt.bar(languages, students_count)
plt.xlabel("Язык программирования")
plt.ylabel("Количество учеников")
plt.title("На каких языках уже программировали ученики")
plt.show()
```
