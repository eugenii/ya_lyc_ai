### Анализ графиков


```python
# Необходимые импорты
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Получение данных и обзор
df = pd.read_csv("../kinopoisk-top250.csv")
df.info()
```

### Гистограмма

Гистограмма plt.hist показывает распределение числовых значений. Например: возрасты людей в группе, оценки за контрольную, рейтинги фильмов.

```python
# Представим, что у нас есть оценки по математике у десяти учеников 
# и мы хотим визуализировать эти данные:

grades = [5, 4, 5, 3, 4, 5, 4, 5, 3, 4]
plt.hist(grades, bins=5) # bins — сколько "корзинок" (интервалов) сделать
plt.xlabel("Оценка")
plt.ylabel("Количество учеников")
plt.title("Распределение оценок по математике")
plt.show()
```

### Столбчатая диаграмма (bar chart)

Столбчатая диаграмма показывает сравнение категорий: страны, предметы, классы, любимые языки программирования.


```python
# Представим, что мы спросили у группы, на каких языках они уже пробовали писать код:

languages = ["Python", "C++", "Java", "JavaScript"]
students_count = [15, 4, 3, 6]
# Построим столбчатую диаграмму plt.bar:

plt.bar(languages, students_count)
plt.xlabel("Язык программирования")
plt.ylabel("Количество учеников")
plt.title("На каких языках уже программировали ученики")
plt.show()
```

_Для чего нам может пригодиться:_

сравнить количество фильмов по жанрам или странам;

сравнить средний рейтинг по годам или по категориям;

показать топ-5 чего угодно (городов, товаров, режиссёров).

Отличие от гистограммы: в гистограмме категории образуются автоматически из числовых 
интервалов (например, 0–10, 10–20, 20–30), а в столбчатой диаграмме категории заданы явно.

### Линейный график

Линейный график удобен, когда мы хотим показать изменение чего-то в хронологическом порядке: по времени, по шагам эксперимента, по годам.

```python
# Представим, что мы записали, сколько задач по программированию 
# решили ученики курса за четыре недели:

weeks = [1, 2, 3, 4]
tasks_solved = [5, 8, 10, 12]
plt.plot(weeks, tasks_solved)
plt.xlabel("Неделя")
plt.ylabel("Количество решённых задач")
plt.title("Прогресс по количеству решённых задач")
plt.grid(True) # сетка, чтобы проще было ориентироваться
plt.show()
```

_Как нам может пригодиться:_

показать, как меняется средний рейтинг фильмов по годам;

показать динамику количества фильмов в топе по годам;

отслеживать прогресс (например, сколько задач решает ученик по неделям).

### Гистограмма: распределение рейтингов
Вернёмся к нашему датасету про фильмы и, для того чтобы проанализировать рейтинги, построим гистограмму:


```python

plt.figure(figsize=(8, 5))
plt.hist(df["rating_ball"], bins=15)
plt.xlabel("Рейтинг фильма")
plt.ylabel("Количество фильмов")
plt.title("Распределение рейтингов топ-250")
plt.show()
```
