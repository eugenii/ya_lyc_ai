#### 5 Самый популярный актёр

```python
# Актёры из ячейки "actors"
from collections import defaultdict


dict_of_actors = defaultdict(int)

def get_actors(col, res=dict_of_actors):
    actors = col.split('; ')
    if actors and actors[-1] == '...':
        actors.pop()
    for actor in actors:
        res[actor] += 1
    
    return len(actors)


df['actors_column'] = df['actors'].apply(get_actors, res=dict_of_actors)
print(dict(dict_of_actors))

# Самый популярный в итогее актёр:
max_count = max(dict_of_actors.values())
name = sorted(dict_of_actors.items(), key=lambda x: x[1], reverse=True)[0][0]
print(name)
```

#### 6 Сценаристы и рейтинги их лучших фильмов

```python
# Сколько сценаристов максимум?
def max_num_get_writers(col):
    return len(col.split('; '))

df['num_writers'] = df['screenwriter'].apply(max_num_get_writers)
print(df['num_writers'].max())

# многоточие - признак большого количества сценаристов, т. е. максимум - трое
```

```python
# Составим словарь со сценаристами - и их колчеством
from collections import defaultdict

screen_writers = defaultdict(int)

def get_writers(col, res=screen_writers):
    writers = [w.strip() for w in col.split('; ') if w != '...']
    for writer in writers:
        res[writer] += 1
    return len(writers)  # Возвращаем количество сценаристов в строке    

# Список со сценаристами, с максимальным количеством фильмов

df['num_of_writers'] = df['screenwriter'].apply(get_writers)
top_screen_writers = list(sorted(screen_writers.items(), key=lambda x: (-x[1], x[0]))[:5])
top_screen_writers = [i[0] for i in top_screen_writers]
display(top_screen_writers)

# Создадим датафрейм по топ 5 сценаристам и их фильмам
# Сначала отметим строки, где присутствуют топ 5 сценаристы


def is_in_top_screen_writers(col):
    top_screen_writers = [w.strip() for w in col.split('; ') if w != '...']
    for writer in top_screen_writers:
        if writer in top_screen_writers:
            return True
    return False

df['is_in_top_screen_writers'] = df['screenwriter'].apply(is_in_top_screen_writers)
display(df.tail(20))
```

```python
# Добавим ВСЕХ сценаристов как столбцы
def put_all_writers(col, st=0):
    wrs = col.split('; ')
    if wrs[-1] == '...':
        wrs.pop()
    while len(wrs) < 3:
        wrs.append('')
    return wrs[st]


writers = ['first_writer', 'second_writer', 'third_writer']
for i in range(len(writers)):
    df[writers[i]] = df['screenwriter'].apply(put_all_writers, st=i)

print(df.head(20))
```
