### 03 Топ-5 режисёров по количеству фильмов

```python
# ИМпорт библиотек и данных
import pandas as pd

df = pd.read_csv('../kinopoisk-top250.csv')
```

```python
# обзор данных

df.info()
display(df.head(1))
```

```python
# Подсёт фильмов у режисёров

directors = df.groupby('director')['movie'].count()
directors = directors.sort_values(ascending=False)
print(directors.head(7))
```

```python
directors_2 = df['director'].value_counts()
display(directors_2.head(7))
```

```python
top_5_directors = df['director'].value_counts().head(5).index.tolist()
result = ','.join(top_5_directors)
print(result)
```
