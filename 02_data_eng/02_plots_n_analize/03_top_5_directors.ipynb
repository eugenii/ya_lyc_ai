### Топ-5 режиссёров

```python
# Импорт библиотек

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Импорт данных

df = pd.read_csv('../kinopoisk-top250.csv')
```

```python
# Подсчитываем количество фильмов у режиссёров

director_counts = df['director'].value_counts()

director_counts = director_counts.sort_index().sort_values(ascending=False)
# Выбираем топ режиссеров
top_directors = director_counts.head(10)

# Смотрим результат
print(top_directors, top_directors.index)

# for director in top_directors.index:
#     print(director.split()[-1], end=',')

# Аналогичное действие в pandas:
surnames = top_directors.index.str.split().str[-1]

# Объединяем фамилии через запятую и выводим результат
result = ','.join(surnames)
print(result)
```
