```python
# Импорты
import pandas as pd
```

```python
#  Загрузка данных.

df = pd.read_csv('../kinopoisk-top250.csv')
df.info()
```

```python
line = 'Австрия, Бельгия, Болгария, Венгрия, Германия, Греция, Дания, Ирландия, Испания, Италия, Кипр, Латвия, Литва, Люксембург, Мальта, Нидерланды, Польша, Португалия, Румыния, Словакия, Словения, Финляндия, Франция, Хорватия, Чехия, Швеция, Эстония'
es_countries = line.split(', ')
print(es_countries)
print(len(es_countries))
```

```python
df['es_countries'] = df['country'].isin(es_countries)
print(df['es_countries'].head(2))
es_countries_df = df[df['es_countries'] == True]
es_countries_df.head(3)
es_countries_ratings = es_countries_df.groupby('country')['rating_ball'].mean()
es_countries_ratings.head(3)
es_countries_avg_ratings = df[df['es_countries'] == True]['rating_ball'].sum() / len(df[df['es_countries'] == True])
print(round(es_countries_avg_ratings, 3)) 
```
