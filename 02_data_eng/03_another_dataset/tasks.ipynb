```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../mls_beauty_data.csv')
df.info()
```

```python
# Переименовываем для удобства работы
df = df.rename(columns={
    'Base Salary': 'base_salary',
    'G/90min': 'goals',
    'A/90min': 'assists',
    'Club.y': 'club'
})

print("Обновленные столбцы:")
print(df.columns.tolist())
print(df.info())
```

```python
# 01 Идеал красоты.

max_beauty = df['beauty'].max()
min_beauty = df['beauty'].min()
print(f"Максимальная красота: {round(max_beauty, 6)}")
print(f"Минимальная красота: {min_beauty.round(6)}")
```

```python
# 02  Счёт звёзд. (designated)

stars_count = df[df['designated'] == 1]['designated'].count()
print("Количество звёзд: {}".format(stars_count))
```

```python
# 03 Голевая эффективность.

stars = df[df['designated'] == 1]
non_stars = df[df['designated'] == 0]
avg_goals_stars = stars['goals'].mean()
avg_goals_non_stars = non_stars['goals'].mean()
display(avg_goals_stars, avg_goals_non_stars)
avg_stars = stars['goals'].sum() / len(stars)
avg_non_stars = non_stars['goals'].sum() / len(non_stars)
print(round(float(avg_stars), 3), float(avg_non_stars.round(3)))
```

```python
# 04 Итоговый доход.

# Требуется вывести 4 числа, среднее значение и медиану base_salary, 
# среднее значение и медиану total_compensation

total_compensation = df['base_salary'] + df['bonus']
base_mean, base_median = df['base_salary'].mean(), df["base_salary"].median()
total_mean, total_median = total_compensation.mean(), total_compensation.median()
print(round(base_mean), ...)

print(*map(round, [base_mean, base_median, total_mean, total_median]))
```

```python
# 05 Клубные бюджеты.

# 1. Считаем количество и среднее
stats = df.groupby('club')['base_salary'].agg(['count', 'mean'])

# 2. Сортируем: сначала по количеству (убывание), 
# затем по названию клуба (возрастание) для однозначности
top_5 = stats.sort_values(by=['count', 'club'], ascending=[False, True]).head(5)

# 3. Округляем (математически: +0.5 и в int) и выводим
result = (top_5['mean'] + 0.5).astype(int)
print(*result)
```

```python
# 6 Контраст профилей

# 1. Список колонок в нужном порядке
cols = ['base_salary', 'beauty', 'assists', 'goals', 'age', 'bonus']

# 2. Группируем по статусу и считаем среднее
grouped = df.groupby('designated')[cols].mean()

# 3. Вычисляем разницу: "звезды" (индекс 1) минус "обычные" (индекс 0)
diff = grouped.loc[1] - grouped.loc[0]

# 4. Выводим результат через пробел с округлением до 3 знаков
print(*diff.round(3))
```
