```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Загружаем датасет
df = pd.read_csv("../mls_beauty_data.csv")
```

```python
# Осмотр данных.
df.info()

# Смотрим на первые строки
print(df.head())

# Проверяем названия столбцов
print("Столбцы в датасете:")
print(df.columns.tolist())
```

```python
# Переименовываем для удобства работы
df = df.rename(columns={
    'Base Salary': 'base_salary',
    'G/90min': 'goals',
    'A/90min': 'assists',
    'Club.y': 'club'
})

print("Обновленные столбцы:")
print(df.columns.tolist())
```

```python
# Количество пропусков по столбцам
missing = df.isna().sum()
print(missing[missing > 0])

# Процент пропусков
missing_percent = (df.isna().sum() / len(df) * 100).round(2)
print(missing_percent[missing_percent > 0])
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['beauty'], bins=12, edgecolor='black')
plt.xlabel('Индекс красоты (симметрия лица)')
plt.ylabel('Количество игроков')
plt.title('Распределение индекса красоты')
plt.grid(alpha=0.3)
plt.show()

print(f"Минимальная красота: {df['beauty'].min():.3f}")
print(f"Максимальная красота: {df['beauty'].max():.3f}")
print(f"Стандартное отклонение: {df['beauty'].std():.3f}")
```
