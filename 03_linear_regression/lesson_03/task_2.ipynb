```python
import pandas as pd
import numpy as np
```

```python
y_test = pd.read_csv("y_test.csv")
y_train = pd.read_csv("y_train.csv")
print(y_test)
y_test_x = y_test['Sales'].values
print(y_test_x)
y_test_y = np.array(y_test_x)
```

```python
baseline = y_train.mean()
```

```python
count = int(input())
def mse_current():
    data = np.array([float(i) for i in input().split()]).reshape(-1, 1)

    return ((y_test['Sales'] - data) ** 2).mean()

for i in range(count):
    res = mse_current()
    if float(res) > baseline['Sales']:
        print(False)
        continue
    print(True)
```
