```python
import pandas as pd
import numpy as np
```

```python
y_test = pd.read_csv("y_test.csv")
y_train = pd.read_csv("y_train.csv")
print(y_test.info(), y_train.info())
```

```python
baseline = y_train.mean()
```

```python
count = int(input())
def mse_current():
    data = np.array([float(i) for i in input().split()]).reshape(-1)
    return ((y_test - data) ** 2).mean(axis=1)

for i in range(count):
    if mse_current() > baseline:
        print(False)
        continue
    print(True)
```
