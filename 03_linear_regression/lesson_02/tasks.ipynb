```python
# 2 MSE
# [float(input() for _ in input().split())]

real = np.array([float(i) for i in input().split()])
predict = np.array([float(i) for i in input().split()])
diff = real - predict
diff_square = diff ** 2
mse = diff_square.mean()
print("MSE: {}".format(mse.round(2)))
```

```python
# 3 MAE

count = int(input())
real = np.array([float(i) for i in input().split()])
# N = int(real.shape[0])

def get_mae(real, predict):
    """ArithmeticError.
    args: real and predict vectors.
    return: mae.
    """
    mae = np.abs(real - predict).mean()
    return mae

mae = None
for _ in range(count):
    predict = np.array([ float(i) for i in input().split()]) 
    if mae is None:
        mae = get_mae(real, predict)
        continue
    mae = min(mae, get_mae(real, predict))
print(mae.round(2))

'''
3
25.2 98.7 69.4 0.7 39.6
25.2 96.8 69.8 2.0 38.5
26.8 96.5 71.2 -0.3 39.5
26.5 99.7 67.5 0.9 42.1
'''
```
