```python
# 3

n = int(input())
data = []
for _ in range(n):
    try:
        item = float(input())
    except:
        item = np.nan
    data.append(item)

data = np.array(data)
median = np.nanmedian(data)

data[np.isnan(data)] = median
print(*data)

```
