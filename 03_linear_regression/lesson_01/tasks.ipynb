```python
# Импорты
import numpy as np
from scipy import stats
```

```python
# 1

a = np.array([3, 2, 1, 0, 2, 0, 3, 2, 2, 1, 3])

mode_result = stats.mode(a)
print("Мода: {}, количество элементов: {}".format(mode_result[0], mode_result[1]))
```

```python
# 2
line = input()
t = input()
hours = np.array([int(i) for i in line.split()])
prices = np.array([int(i) for i in t.split()])

salary = np.dot(hours, prices)
print(salary)
```
