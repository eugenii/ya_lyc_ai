```python
# 02 Эксцентриситет.

def eccentricity(a, b):
    e = (a * a - b * b) ** 0.5 / a
    print(e)

eccentricity(149.597, 149.575)
```

```python
# 03 Средний балл.


def average(marks):
    avg = sum(marks) / len(marks)
    print(avg)


average([1, 2, 3, 4])
```

```python
# 04 Границы.


def borders(a: int, b: int, c: int) -> None:
    """Вывести расположение точки по отношению к прямоугольнику.

    : args 1, 2 координаты точек a, b
    : arg 3 координаты проверяемой точки c.
    : return None
    """
    if c[0] >= a[0] and c[1] <= a[1] and c[0] <= b[0] and c[1] >= b[1]:
        if c[0] == a[0] or c[0] == b[0] or c[1] == a[1] or c[1] == b[1]:
            print('AT THE EDGE')
        else:
            print('INSIDE')
    else:
        print('OUTSIDE')

    
borders((2, 9), (5, 1), (3, 4))
borders((2, 9), (5, 1), (13, 4))
borders((2, 9), (5, 1), (2, 4))
```
