```python
# 04 Границы.


def borders(a: int, b: int, c: int) -> None:
    """Вывести расположение точки по отношению к прямоугольнику.

    : args 1, 2 координаты точек a, b
    : arg 3 координаты проверяемой точки c.
    : return None
    """
    if c[0] >= a[0] and c[1] <= a[1] and c[0] <= b[0] and c[1] >= b[1]:
        if c[0] == a[0] or c[0] == b[0] or c[1] == a[1] or c[1] == b[1]:
            print('AT THE EDGE')
        else:
            print('INSIDE')
    else:
        print('OUTSIDE')

    
borders((2, 9), (5, 1), (3, 4))
borders((2, 9), (5, 1), (13, 4))
borders((1, 3), (4, 0), (4, 4))
```

Задача 5 Конь


```python
# 5 Конь

def horse(p1: str, p2: str) -> bool:
    """
    Вычисляет, может ли конь попасть из p1 в p2
    :arg1 p1: начальная позиция
    :arg2 p2: конечная позиция

    :return: True, если конь может попасть из p1 в p2, иначе False
    """
    line1 = 'abcdefgh'
    line2 = '12345678'
    p1 = (line1.index(p1[0]), line2.index(p1[1]))
    p2 = (line1.index(p2[0]), line2.index(p2[1]))
    if p1 == p2:
        print('False')
    elif p1[0] == p2[0] or p1[1] == p2[1]:
        print('False')
    elif abs(p1[0] - p2[0]) == 2 and abs(p1[1] - p2[1]) == 1:
        print('True')
    elif abs(p1[0] - p2[0]) == 1 and abs(p1[1] - p2[1]) == 2:
        print('True')
    else:
        print('False')

    return abs(p1[0] - p2[0]) == abs(p1[1] - p2[1]) or (p1[0] == p2[0] or p1[1] == p1[0])


horse('d4', 'b3')    # True

horse('a1', 'c1')    # False
    
```

```python
# 6 Прямая через 2 точки (y = k * x + b)

def line(x1: int, y1: int, x2, y2) -> None:
    k = (y2 - y1) / (x2 - x1)
    b = y1 - k * x1
    if k == 0:
        k = '0.0'
    elif k < 0:
        k = str(round(k, 2))
    else:
        k = str(round(k, 2))
    
    if b < 0:
        b = ' - ' + str(abs(round(b, 2)))
    else:
        b = ' + ' + str(round(b, 2))
    line = 'y = ' + k + ' * ' + 'x' + b
    print(line)


line(0, 4, -2, 4)
line(0, -1, 4, 0)
```
