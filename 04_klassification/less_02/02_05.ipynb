```python
# Импорты
import pandas as pd
```

```python
data = pd.read_csv('titanic.csv', sep=',')
print(data.info())
```

```python
# Данные о выживших пассажирах.
data_surv = data[data['Survived'] == 1]
print(data_surv.head())

# Средний возраст выживших.

print(data_surv['Age'].mean().round(2))
```

```python
# Самая большая семья на титанике
# Берем столбец Name и разделяем по запятой и находим длину
data['family'] = data['Name'].str.split(', ')
data['family_size'] = data['family'].str.len()
max_family_size = data['family_size'].max()
print(data.head())
print(max_family_size)
```

```python
# Сколько женщин было на борту?
print(data['Sex'].value_counts())
```

```python
# В каком порту больше всего село?
print(data['Embarked'].value_counts())
```
